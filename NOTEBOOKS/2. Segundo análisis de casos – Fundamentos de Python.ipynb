# Problema #1: Calculadora de propina inteligente

## Problema

El Problema a resolver es cuanta propina se va a dar y sumarlo al monto final, pidiendo al usuario los datos de cuanto es el total de la cuenta y cual es el porcentaje de la propina.

## Entrada

La entrada en este caso sera de dos valores, la cuenta como flotante y el porcentaje de propina como flotante tambien.

## Proceso

Evaluaremos que el monto de la cuenta no sea 0 o menor. Una vez hecho esto procederemos a validar si la propina es 0 o un numero positivo.
Una vez las validaciones esten concluidas procederemos a sacar el valor numerico de la propina y sumarlo a la cuenta.
Para eso usaremos la operacion de matematica diviendo entre 3 cuenta por propina divido entre 100.

## Salida esperada

La salida sera una cadena que indique el monto de la cuenta el monto de la propina y el total.


## Pruebas manuales

| Escenario | Datos de entrada               |            Resultado esperado                 |         Resultado obtenido             |
|-----------|--------------------------------|-----------------------------------------------|----------------------------------------|
| Caso normal | Cuenta 1500 Propina 5% | Total 1575 Propina 75 | Total 1575 Propina 75 |
| Caso límite | Cuenta 1000000000 Propina 15% | Cuenta 1150000000.00. Propina 150000000.00 | Cuenta 1150000000.00. Propina 150000000.00 |
| Caso inválido | Cuenta -0 | except | except |


```python
try:
    cuenta = float(input("Por Favor Digite el Monto de la Cuenta: ")) #Pedimo el monto de la Cuenta
    propina = float(input("Por Favor Digite el Monto que desea dejar de Propina ✌️: ")) #Monto que desea dejar de propina
    
    if cuenta <= 0:  #Verifico que la Cuenta no sea 0 o menos
        print("La Cuenta debe ser mayor a cero.")
    elif propina < 0: # Verifico que la propina no sea Negativa, el usuario puede descidir no dejar propina
        print("La propina no puede ser negativa.")
    else:
        total = cuenta + (cuenta * propina / 100) #A la cuenta le sumo el valor de la propina para el total         
        print(f"El total de tu cuenta es: {total:.2f}.\n El Subtotal sería: {cuenta:.2f} y el monto de la propina: {(cuenta * propina / 100):.2f}")    
except ValueError:
    print("Por favor los montos digitados deben ser de numeros.")
```

# Problema #2: Detector de número par/impar

## Problema

El Usuario necesita identificar si el numero es par o impar. Se le solicitara un numero al usuario y el sistema devera devolver si es par o impar.

## Entrada

En esta caso la entrada sera el numero a evaluar que debe ser un numero.

## Proceso
En este caso evaluaremos si el numero ingresado por el usuario es par o impar. Para hacer esto usaremos el operador % de Python que nos dira si el resto de la division por el numero que digitemos en el codigo a continuacion siendo el resultado 1 o 0.

## Salida esperada

La salida esperada en este caso es una cadena de string que nos diga si el numero digitado es par o impar.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | 8  | par | par |
| Caso límite | 89742159 | impar | impar |
| Caso inválido | 89.5 | except  | except |



```python
try:
    numero = int(input("Por Favor Digite el numero a evaluar: ")) #Pedimos el numero a evaluar
    resto = numero % 2 # Asigo el resto de la division entre dos a la variable resto 0 si es par 1 si es impar
    print(resto)
    if resto:  #Verifico si resto es 1 o Verdadero, en este caso es impar
        print(f"El numero: {numero}, es un numero impar.")
    else: #Si en la validacion anterior es falsa el valor es 0 o negativo siendo el numero impar
        print(f"El numero: {numero}, es un numero par.")    
except ValueError:
    print("Por favor digitar solamente numeros enteros.")
```

# Problema #3: Clasificador de edad

## Problema

Recibe la edad de la persona y clasifica en Menor menos de 18 años, Adulto de 18 a 64 y Adulto Mayor, mayor de 65

## Entrada

En este caso la edad de la persona sera el dato de entrada.

## Proceso

En este caso vamos a tomar el valor de entrada("Edad") y asignarlo a una variable llamada edad.
Con esta variable realizaremos varias comparaciones para ver si la edad es mayor a 0, si la edad es mayor que 0 y menor de 18 cayendo en la categoria de Menor de Edad. Si por el contrario el valor de la edad esta en el rango de 18 a menos de 65 la edad sera Mayor de Edad. Ahora si la edad es mayor a 65 la persona es considerada Adulto Mayor.

## Salida esperada

La salida esperada es una cadena de texto indicando que segun la edad digitada indicando la edad eso se hace menor de edad, mayor de edad o adulto mayor segun sea el caso.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | 20 | Mayor de edad | Mayor de edad |
| Caso límite | 99 | Adulto Mayor | Adulto Mayor |
| Caso inválido | trece | except | except |


```python
try:
    edad = int(input("Por Favor Digite su Edad : ")) #Pedimos la edad de la Persona
    if edad <= 0:  #Verifico que la Edad no sea 0 o menos
        print(f"La edad digitada : {edad} Recuerda que la edad debe ser mayor a 0")
    elif edad > 0 and edad < 18: # Verifico que la edad sea entre 0 o mayor y menor 18
        print(f"Su edad es: {edad} eso te hace Menor de edad ")
    elif edad >= 18 and edad < 65: # Verifico que la edad sea entre 18 o mayor y menor a 65
        print(f"Su edad es: {edad} eso te hace Mayor de edad ")
    else:  # VAqui no se verifica la edad que que si no cubrio las condiciones anteriores es Adulto Mayor
        print(f"Su edad es: {edad} eso te hace Adulto Mayor de edad ")    
except ValueError:
    print("La edad debe ser digitada en numeros positivos.")
```
