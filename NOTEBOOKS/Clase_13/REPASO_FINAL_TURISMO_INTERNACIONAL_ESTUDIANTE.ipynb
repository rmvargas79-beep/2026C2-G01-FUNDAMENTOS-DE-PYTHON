```python
df1 = (df[df["anio"] == 2019])
df2 = (df[df["anio"] == 2020])
df3 = pd.merge(df1,df2,on="pais",suffixes=("_2019", "_2020"))
df3["caida_porcentual"] = ((df3["llegadas_miles_2019"] - df3["llegadas_miles_2020"]) / df3["llegadas_miles_2019"]) * 100  
df3.sort_values(by="caida_porcentual", ascending=False, inplace=True)

df3
```

```python
df_2019 = (df[df["anio"] == 2019].set_index('pais')['llegadas_miles'])
df_2020 = (df[df["anio"] == 2020].set_index('pais')['llegadas_miles'])

df_impacto = pd.DataFrame({
    '2019': df_2019,
    '2020': df_2020
})

df_impacto['caida_porcentual'] = ((df_impacto['2019'] - df_impacto ['2020']) / df_impacto['2019']) * 100

display(df_impacto.sort_values(by='caida_porcentual', ascending= False))

df_impacto["caida_porcentual"].nlargest(3).round(2)
```
