```python
# TODO CÓDIGO 2: lista de diccionarios, ciclo, condición y función resumen_region.
# Deben quedar visibles: el conteo y las dos llamadas de prueba de resumen_region.

registros = df.to_dict(orient='records')
print(registros)
cantidad_paises = 0
promedio_region = 0
for registro in registros:
    if registro["puntuacion_bienestar"] >= 6:
        cantidad_paises += 1
    
print(f"Paises con Puntuacion de Bienestar mayor a 6.0: {cantidad_paises}")

def resumen_region(registros, region):
    # Recorra los registros y acumule solo los del país indicado.
    cantidad_paises = 0
    sum_bienestar = 0
    prom_bienestar = 0
    if region not in [registro["region"].lower() for registro in registros]:
        print(f"La Region {region} ingresada no se encuentra en los registros.")
    else:
        for registro in registros:
            if registro["region"].lower() == region:
                cantidad_paises += 1
                sum_bienestar = sum_bienestar + (registro["puntuacion_bienestar"])
                
        prom_bienestar = (sum_bienestar / cantidad_paises)
        print(f"La Region {region} con un total de {cantidad_paises} paises tiene un promedio de bienestar de {prom_bienestar}.")
        return
    
    
region = input("Ingrese la region que desea consultar: ").strip().lower()
resumen_region(registros, region)
```

```python
# TODO CÓDIGO 3: promedio, diferencia_promedio, filtros, isin(...), groupby(...).agg(...) y nlargest(...).
# Deben quedar visibles: promedio_muestra, df_region, regiones_comparadas, df_comparacion, resumen_por_region y top_5.
#Regiones Elegidas Asia Oriental, Asia Sudoriental
#Quiero saber que region si al A o B tiene mas paises que superen el indice de 6.0 en Bienestar y cuantos paises de cada region lo superan.

promedio_muestra = (df["puntuacion_bienestar"].sum()/ len(df))
print(f"El Promedio de la muestra es: {promedio_muestra:.3f}")

columnas_a = ["pais", "region", "puntuacion_bienestar"]
columnas_b= ["pais", "region", "puntuacion_bienestar", "Diferencia_Promedio"]

print(f"\nLa Puntuacion de Bienestar de la Region A es: ")
region_a = (df[df["region"] == "Asia Oriental"])
region_a["Diferencia_Promedio"] = (region_a["puntuacion_bienestar"] - promedio_muestra)
display(region_a[columnas_b])

print(f"\nLa Puntuacion de Bienestar de la Region B es: ")
region_b = (df[df["region"] == "Asia Sudoriental"])
region_b["Diferencia_Promedio"] = (region_b["puntuacion_bienestar"] - promedio_muestra)
display(region_b[columnas_b])

print(f"\nLa Comparacion entre las Regiones es: ")
regiones_comparadas = ["Asia Oriental", "Asia Sudoriental"]
df_comparacion = df[(df["region"].isin(regiones_comparadas)) & (df["puntuacion_bienestar"] >= promedio_muestra)]
display(df_comparacion[columnas_a])


print(f"\nEl Resumen de las Regiones es: ")
resumen_regiones = df.groupby("region").agg(promedio_bienestar=("puntuacion_bienestar", "mean") , cantidad_paises=("pais", "count"))
display(resumen_regiones)

print(f"\nLos 5 paises a nivel mundial con mejor puntuacion de Bienestar son: ")
top_5= df.groupby("pais")["puntuacion_bienestar"].sum().nlargest(5)
top_5

```
